## Task 02 - Image Encryption Tool

**Objective:**
Develop a simple image encryption tool using pixel manipulation.
The tool allows users to encrypt and decrypt images using a basic mathematical operation on pixel values.

```python
from PIL import Image
```

```python
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
from IPython.display import display

image = Image.open(next(iter(uploaded)))
display(image)
```

```python
def encrypt_image(image, key):
    image = image.convert("RGB")
    pixels = image.load()

    for x in range(image.width):
        for y in range(image.height):
            r, g, b = pixels[x, y]

            r = (r + key) % 256
            g = (g + key) % 256
            b = (b + key) % 256

            pixels[x, y] = (r, g, b)

    return image
```

```python
key = 50
```

```python
encrypted_image = encrypt_image(image.copy(), key)
display(encrypted_image)
```

```python
encrypted_image.save("encrypted_image.png")
```

```python
files.download("encrypted_image.png")
```

```python
def decrypt_image(image, key):
    return encrypt_image(image.copy(), -key)
```

```python
decrypted_image = decrypt_image(encrypted_image, key)

display(decrypted_image)
```

```python
decrypted_image.save("decrypted_image.png")
```

```python
files.download("decrypted_image.png")
```

## Conclusion

This project demonstrates a simple image encryption and decryption technique using pixel manipulation. A mathematical operation is applied to the RGB values of each pixel using a user-defined key. The original image can be restored by applying the reverse operation during decryption.
